```python
from pysr import PySRRegressor
import numpy as np
```

```python
rng = np.random.default_rng(0)
```

Modeling:
$$ y = y_0 + v_y t + 0.5 * g t^2 $$

```python
def equation(y0, v0, t):
	return y0 + v0 * t - 0.5 * 9.81 * t**2

N = 50

y0 = rng.uniform(-30, 30, N)
v0 = rng.uniform(1, 10, N)
t = rng.uniform(0, 100, N)

y = equation(y0, v0, t)

# noise
eps = rng.normal(0, 0.1, N)
y += eps

y.shape
```

```python
model = PySRRegressor(
	binary_operators=["+", "-", "*", "^"],
	maxsize=20,
	niterations=100, # search effort / timeout
)

X = np.column_stack([y0, v0, t])
model.fit(X, y, variable_names=["y0", "v0", "t"])
```

```python
print(model.equations_[["complexity", "loss", "equation"]])
```

```python
print(model.sympy()) 
```

## Modeling:
$$x(t) = A\, e^{-\frac{b}{2m}t}\cos\!\left(t\sqrt{\tfrac{k}{m} - \tfrac{b^2}{4m^2}}\right)$$

```python
def equation(X):
	A, b, k, t, m = X
	return A * np.exp(-0.5 * t * b / m) * np.cos(t * np.sqrt(k/m - b**2 / (4 * m**2)))

N = 300

A = rng.uniform(-30, 30, N)
b = rng.uniform(0.1, 10, N)
k = rng.uniform(0.1, 10, N)
t = rng.uniform(0, 100, N)
m = rng.uniform(0, 100, N)

X = (A, b, k, t, m)
y = equation(X)

# noise
eps = rng.normal(0, 0.1, N)
y += eps

X = np.column_stack(X)
X.shape, y.shape
```

```python
model = PySRRegressor(
	binary_operators=["+", "-", "*", "^", "/"],
	unary_operators=["exp", "cos", "sqrt"],
	maxsize=35,
	niterations=300, # search effort / timeout
	populations=40,
)
mask = ~np.isnan(X).any(axis=1) & ~np.isnan(y)
model.fit(X[mask], y[mask], variable_names=["A", "b", "k", "t", "m"])
```

```python
print(model.sympy()) 
```
